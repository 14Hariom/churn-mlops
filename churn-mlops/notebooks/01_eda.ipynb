```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/telco_churn.csv")
df.head()
```

## 1. How big is the data and what are the columns?

```python
print("Rows, columns:", df.shape)
df.info()
```

## 2. Data quality problems
`TotalCharges` should be a number, but the raw file stores it as text with blanks.

```python
print(df.isnull().sum().sum(), "missing values (real NaNs)")
print("Blank TotalCharges:", (df["TotalCharges"].astype(str).str.strip() == "").sum())
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)
```

## 3. What percentage of customers churn?
If the classes are very unbalanced, **accuracy alone is a misleading metric** - look at recall and F1.

```python
churn_rate = (df["Churn"] == "Yes").mean()
print(f"Churn rate: {churn_rate:.1%}")
df["Churn"].value_counts().plot(kind="bar", title="Churn counts", rot=0)
plt.show()
```

## 4. Do month-to-month customers churn more?

```python
rate = df.groupby("Contract")["Churn"].apply(lambda s: (s == "Yes").mean())
rate.plot(kind="bar", title="Churn rate by contract type", rot=0)
plt.ylabel("Churn rate")
plt.show()
```

## 5. Do new customers leave more often?

```python
df[df["Churn"] == "Yes"]["tenure"].plot(kind="hist", bins=30, alpha=0.6, label="Churned")
df[df["Churn"] == "No"]["tenure"].plot(kind="hist", bins=30, alpha=0.6, label="Stayed")
plt.title("Tenure (months) by churn")
plt.legend()
plt.show()
```

## 6. Do higher bills mean more churn?

```python
df.boxplot(column="MonthlyCharges", by="Churn")
plt.title("Monthly charges by churn")
plt.suptitle("")
plt.show()
```
